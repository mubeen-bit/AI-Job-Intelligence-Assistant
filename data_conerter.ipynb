```python
import pandas as pd
import os
import re

# Read CSV
df = pd.read_csv("/Users/syedmubeen/projects/job_skills_needed/job_title_des.csv")

# Check column names
print(df.columns)

# Create output folder
output_folder = "markdown_jobs"
os.makedirs(output_folder, exist_ok=True)

# Clean filename function
def clean_filename(name):
    name = re.sub(r'[\\/*?:"<>|]', "", name)
    return name.replace(" ", "_")

# Generate markdown files
for index, row in df.iterrows():

    job_title = str(row["Job Title"])
    description = str(row["Job Description"])

    filename = clean_filename(job_title) + ".md"
    filepath = os.path.join(output_folder, filename)

    markdown_content = f"""# {job_title}

## Job Description

{description}
"""

    with open(filepath, "w", encoding="utf-8") as f:
        f.write(markdown_content)

print("Markdown files created successfully!")
```

```python
import csv
import io
import os

# Your CSV data string
csv_data = """id,title,description
1,Django Developer,"PYTHON/DJANGO (Developer/Lead) - Job Code(PDJ - 04)..."
2,Machine Learning,"Data Scientist (Contractor)..."
...""" # (I have truncated the string here for brevity; paste your full string)

def convert_csv_to_md(csv_text, output_folder="job_descriptions_md"):
    # Create the folder if it doesn't exist
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    
    # Use io.StringIO to treat the string like a file
    f = io.StringIO(csv_text.strip())
    reader = csv.reader(f)
    
    # Skip header row
    headers = next(reader)
    
    count = 0
    for row in reader:
        if len(row) < 3:
            continue
            
        job_id = row[0].strip()
        job_title = row[1].strip()
        job_desc = row[2].strip()
        
        # Clean title for filename (remove slashes/special chars)
        clean_title = "".join(x for x in job_title if x.isalnum() or x in "._- ").replace(" ", "_")
        filename = f"{job_id}_{clean_title}.md"
        filepath = os.path.join(output_folder, filename)
        
        # Structure the Markdown for better RAG indexing
        md_content = f"# Job Title: {job_title}\n\n"
        md_content += f"## Job ID: {job_id}\n\n"
        md_content += f"## Description\n{job_desc}\n\n"
        md_content += "---\n"
        md_content += f"Metadata: ID={job_id}, Role={job_title}"

        with open(filepath, "w", encoding="utf-8") as md_file:
            md_file.write(md_content)
        
        count += 1
        print(f"Created: {filename}")

    print(f"\nSuccess! {count} markdown files generated in '{output_folder}'.")

if __name__ == "__main__":
    # Specify the path to your CSV file here
    input_file_path = "/Users/syedmubeen/projects/job_skills_needed/job_title_des.csv" 
    
    with open(input_file_path, "r", encoding="utf-8") as f:
        csv_text = f.read()
        
    convert_csv_to_md(csv_text)
```
