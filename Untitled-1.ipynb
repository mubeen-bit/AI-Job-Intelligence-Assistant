```python
import requests
from bs4 import BeautifulSoup

ROLE = "Java Developer"

url = f"https://www.indeed.com/jobs?q={ROLE.replace(' ', '+')}"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.text, "html.parser")

jobs = soup.find_all("div", class_="job_seen_beacon")

for idx, job in enumerate(jobs[:5], start=1):

    title = job.find("h2").text.strip()

    snippet = job.find("div", class_="job-snippet")

    description = snippet.text.strip() if snippet else "No description"

    output = f"""
# Job Title: {title}

## Job ID: {idx}

## Description
{description}

---
Metadata: ID={idx}, Role={ROLE}
"""

    print(output)
```
